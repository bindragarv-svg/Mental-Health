```python
top_country = df.Country.value_counts().index[:10].tolist()
top_country
```

```python
def grop(country):
    if country in top_country:
        return country
    else:
        return 'Other'
```

```python
df['Group_country'] = df.Country.apply(grop)
```

```python
skew_column = ['Study_Hours']
standard_col  =['Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
one_hot_col = ['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','Group_country']
```

```python
from sklearn.model_selection import train_test_split
col = skew_column + standard_col + ordinal_col + one_hot_col
x = df[col]
y = df['Mental_Health_Score']
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder,FunctionTransformer
from sklearn.pipeline import Pipeline

skew_pipeline = Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

non_skew_pipeline = Pipeline(steps=[
    ('scale',StandardScaler())
])

ordinal = Pipeline(steps=[
    ('Encoding',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

one_hot = Pipeline(steps = [
    ('Encoding',OneHotEncoder(handle_unknown = 'ignore'))
])
```

```python
from sklearn.compose import ColumnTransformer

col_transform = ColumnTransformer(transformers=[
    ('Skew',skew_pipeline,skew_column),
    ('non_skew',non_skew_pipeline,standard_col),
    ('ordinal_encoding',ordinal,ordinal_col),
    ('one_hot',one_hot,one_hot_col)
])
col_transform
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

lr_pipeline = Pipeline(steps=[
    ('preprocessor',col_transform),
    ('model',LinearRegression())
])
lr_pipeline
```

```python
lr_pipeline.fit(X_train,y_train)
```

```python
y_predict = lr_pipeline.predict(X_test)
y_predict_train = lr_pipeline.predict(X_train)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
r2_score_test = r2_score(y_test,y_predict)
r2_score_train = r2_score(y_train,y_predict_train)
ame= mean_absolute_error(y_test,y_predict)
print(f"the test score is {r2_score_test}")
print(f"the train score is {r2_score_train}")
print(f"the ame score is {ame}")
```

```python
from sklearn.ensemble import RandomForestRegressor

Random_pipeline = Pipeline(steps=[
    ('preprocessor',col_transform),
    ('model',RandomForestRegressor(random_state=42))
])
Random_pipeline
```

```python
Random_pipeline.fit(X_train,y_train)
```

```python
y_predicttest = Random_pipeline.predict(X_test)
y_predicttrain = Random_pipeline.predict(X_train)
r2_score_ra_test = r2_score(y_test,y_predicttest)
r2_score_ra_train = r2_score(y_train,y_predict_train)
amera =mean_absolute_error(y_test,y_predicttest)

print(f"the testing score is {r2_score_ra_test}")
print(f"the training score is {r2_score_ra_train}")
print(f"the ame score is {amera}")
```

Hyper parameter turning  for random forest algorithm we use the randomizedSearch CV

```python
from sklearn.model_selection import RandomizedSearchCV
grid = {
    'model__n_estimators':[100,200,500],
    'model__max_depth':[5,10,15],
    'model__min_samples_split':[2,4,6],
    'model__min_samples_leaf':[1,2,4]
}

mrf = RandomizedSearchCV(
    estimator=Random_pipeline,
    param_distributions=grid,
    n_jobs=1,
    random_state=42,
    n_iter = 15
)
mrf.fit(X_train,y_train)
```

```python
best_ra_model = mrf.best_estimator_
preytest = best_ra_model.predict(X_test)
preytrain = best_ra_model.predict(X_train)
sc_test = r2_score(y_test,preytest)
sc_train = r2_score(y_train,preytrain)
amerg = mean_absolute_error(y_test,preytest)
```

```python
print(f"the testing r2 score is {sc_test}")
print(f"the training r2 score is {sc_train}")
print(f"the ame is {amerg}")
```
